```python
from openai import OpenAI

client=OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENAI_API_KEY"]
)

EMBEDDING_MODEL="openai/text-embedding-3-small"
CHAT_MODEL="openai/gpt-4.1-mini"

print("client initialized success")
```

```python
# Defines a function 'embed_text' that takes a string 'text' and returns a list of floats (the embedding vector).
def embed_text(text: str) -> list[float]:
    # Calls the OpenAI client's 'embeddings.create' method to generate an embedding.
    response = client.embeddings.create(
        # Specifies the embedding model to use.
        model=EMBEDDING_MODEL,
        # Provides the text to be embedded.
        input=text
    )
    # Extracts and returns the embedding vector from the response.
    return response.data[0].embedding
```

```python
# Imports the 'QdrantClient' class for interacting with the Qdrant vector database.
from qdrant_client import QdrantClient
# Imports 'Distance' (for specifying vector comparison metric) and 'VectorParams' (for configuring vector storage) models.
from qdrant_client.models import Distance, VectorParams

# Defines the name of the Qdrant collection (like a table in a traditional database).
COLLECTION_NAME = "long_term_memory"
# Defines the dimensionality of the vectors that will be stored (this matches the output size of 'text-embedding-3-small').
VECTOR_SIZE = 1536

# Initializes an in-memory Qdrant client, meaning the database exists only for the current session.
qdrant = QdrantClient(":memory:")

# Creates a new collection in the Qdrant database.
qdrant.create_collection(
    # Specifies the name of the collection.
    collection_name=COLLECTION_NAME,
    # Configures the vectors within the collection.
    vectors_config=VectorParams(
        # Sets the size (dimensionality) of the vectors.
        size=VECTOR_SIZE,
        # Specifies Cosine similarity as the distance metric for vector comparison.
        distance=Distance.COSINE
    ),
)

# Prints a confirmation that the memory store (Qdrant collection) is ready.
print("Long-term memory store ready.")

```

Store and retrive memory function

```python
# Imports 'PointStruct' for creating data points to be stored in Qdrant.
from qdrant_client.models import PointStruct

# Defines a function to store a piece of text as a memory, taking the text and a unique ID.
def store_memory(text: str, memory_id: int):
    # Generates the embedding vector for the input text using the 'embed_text' function.
    vector = embed_text(text)
    # Inserts or updates a point in the Qdrant collection.
    qdrant.upsert(
        # Specifies the target collection.
        collection_name=COLLECTION_NAME,
        points=[ # Provides a list of points to upsert.
            PointStruct(
                # Assigns the unique ID to the point.
                id=memory_id,
                # Stores the embedding vector.
                vector=vector,
                # Stores the original text as a payload associated with the vector.
                payload={"text": text}
            )
        ]
    )

# Defines a function to retrieve memories similar to a given query, returning 'top_k' results.
def retrieve_memory(query: str, top_k: int = 3) -> list[str]:
    # Generates the embedding for the query text.
    query_vector = embed_text(query)
    # Queries the Qdrant collection for similar vectors.
    search_result = qdrant.query_points(
        # Specifies the collection to search.
        collection_name=COLLECTION_NAME,
        # Provides the query vector.
        query=query_vector,
        # Sets the maximum number of results to return.
        limit=top_k
    )
    # Extracts and returns the original text from the payloads of the top 'k' search results.
    return [hit.payload["text"] for hit in search_result.points]
```

```python
# Stores the first memory about Anupam's profession with ID 1.
store_memory("Anupam works as a Senior Research Analyst.", 1)
# Stores the second memory about Anupam's preference with ID 2.
store_memory("Anupam prefers step-by-step technical explanations.", 2)
# Stores the third memory about Anupam's current learning with ID 3.
store_memory("Anupam is learning agent memory and state management.", 3)

# Prints a confirmation message that the memories have been stored.
print("Memories stored.")
```

```python
# Calls the 'retrieve_memory' function with a question to find relevant memories.
recalled = retrieve_memory("What does Anupam do professionally?")
# Prints a header for the recalled memories.
print("\nRecalled Memories:")
# Loops through each recalled memory.
for r in recalled:
    # Prints each recalled memory, prefixed with a hyphen.
    print("-", r)

```

```python
# Defines a function 'agent_reply' that takes a user question and returns an AI-generated answer.
def agent_reply(user_question: str) -> str:
    # Retrieves relevant memories based on the user's question.
    memory_context = retrieve_memory(user_question)
    # Joins the retrieved memories into a single string, separated by newlines, to form a context block.
    memory_block = "\n".join(memory_context)

    # Calls the OpenAI client's 'chat.completions.create' method to generate a chat completion.
    response = client.chat.completions.create(
        # Specifies the chat model to use.
        model=CHAT_MODEL,
        messages=[ # Provides a list of messages for the chat history.
            # Sets the system role message, instructing the LLM to use the provided memory.
            {"role": "system", "content": "Use the provided memory to answer accurately."},
            # Sets the user role message, including the retrieved memory block and the original user question.
            {"role": "user", "content": f"MEMORY:\n{memory_block}\n\nQUESTION:\n{user_question}"}
        ]
    )
    # Extracts and returns the content of the AI's reply.
    return response.choices[0].message.content

# Calls the 'agent_reply' function with a specific question about Anupam.
answer = agent_reply("Tell me about Rahul Dravid.")
# Prints a header for the agent's response.
print("\nAgent Response:")
# Prints the AI-generated answer.
print(answer)
```
